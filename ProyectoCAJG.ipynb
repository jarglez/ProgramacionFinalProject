# **1. Introducción**

## - ⚠️ Descripción del problema: ¿Qué queremos analizar o predecir?

El infarto agudo de miocardio (IAM) es una de las principales causas de mortalidad en el mundo. Durante la hospitalización, algunos pacientes desarrollan **complicaciones graves y mortales**. La capacidad de identificar los pacientes con riesgo de estas complicaciones desde el ingreso del paciente permitiría **optimizar los recursos clínicos**, dirigir el **esfuerzo terapeútico**, mejorar la toma de decisiones y **reducir la mortalidad hospitalaria**.

En este análisis, buscamos desarrollar un modelo de calsificación que permita identificar qué pacientes tienen mayor riesgo de presentar **complicaciones mortales** tras un infarto.  

## - 🎯 Objetivo del análisis: ¿Qué impacto tendrá este análisis?

El objetivo principal de este estudio es construir un modelo de calsificación de mortalidad y complicaciones mortales basado en datos clínicos y demográficos disponibles en el momento del ingreso, o en las primeras horas de estancia intrahospitalaria del paciente. Este modelo clasificará a los pacientes en una de las siguientes categorías mortales según su evolución:

- **Vivo**: El paciente sobrevive sin complicaciones críticas. (**NO TUVO COMPLICACIONES FATALES, PUDO HABER TENIDO COMPLICACIONES NO FATALES**)
- **Shock cardiogénico** 🫀: Estado de insuficiencia circulatoria grave debido a la incapacidad del corazón para bombear sangre de manera efectiva.
- **Edema pulmonar** 🌊: Acumulación de líquido intersticial en los pulmones, lo que dificulta la oxigenación y puede llevar a insuficiencia respiratoria.
- **Ruptura miocárdica** 💔: Desgarro de la pared del miocardio, lo que puede causar un taponamiento cardíaco y una rápida muerte.
- **Progresión de insuficiencia cardíaca congestiva** 🔄: Deterioro progresivo de la función cardíaca, lo que lleva a una acumulación de líquidos y falla multiorgánica y finalmente muerte.
- **Tromboembolismo** 🩸: Formación de coágulos sanguíneos que pueden desplazarse a órganos vitales, causando infartos cerebrales, pulmonares o sistémicos.
- **Asistolia** ⛔: Ausencia total de actividad eléctrica en el corazón, lo que resulta en un paro cardíaco irreversible.
- **Fibrilación ventricular** ⚡: Ritmo cardíaco caótico que impide que el corazón bombee sangre de manera efectiva, llevando rápidamente a la muerte si no se trata de inmediato.

Este modelo permitiría al personal de salud **identificar de manera temprana** a los pacientes con mayor riesgo de complicaciones y tomar medidas preventivas, como **ajustes en el tratamiento, monitoreo intensivo y estrategias de intervención temprana**. En última instancia, esta herramienta podría **reducir la mortalidad hospitalaria y mejorar los resultados clínicos** en pacientes con infarto agudo de miocardio.

## - ⚠️ Debilidades del Dataset  

A pesar de que la **base de datos de complicaciones del infarto de miocardio** es una fuente valiosa de información clínica, presenta varias limitaciones que deben considerarse al utilizarla en análisis modernos:  

### 📅 **Fecha de recolección de los datos (1992-1995)**  
- Los datos fueron recopilados hace **más de 25 años**, lo que plantea un desafío en cuanto a su **relevancia actual**.  
- Desde entonces, ha habido **avances significativos** en el diagnóstico, tratamiento y manejo del infarto agudo de miocardio, lo que podría hacer que los patrones de complicaciones en la base de datos **no reflejen la realidad médica actual**.  
- Medicamentos, procedimientos como **avancees en reanimación cardiopulmonar**, **detección temprana de isquemia cardiaca**, **dispositivos como desfirbiladores automáticos externos**, **programas como código infarto**, la **angioplastia primaria** y el uso de **stents**, así como el manejo en unidades de cuidados intensivos, han evolucionado y cambiado la **tasa de complicaciones**. Por ejemplo, los datos de esta fuente no clasifican el infarto con el paradigma moderno con o sin elevación del segmento ST, lo que disminuye su utilidad. Otro ejemplo es el tratamiento fibrinolítico que usan son del grupo de la esptreptoquinada, tratamiento que actualmente no está en uso.
- A pesar de contar con documentos que describen bien las variables de carece de la descripción metodológica de como se recolectaron los datos y la razón por lo que existen valores nulos o perdidos.  

### 🌍 **Ubicación geográfica: Rusia en un periodo de crisis**  
- La base de datos fue recolectada en **Krasnoyarsk, Rusia**, lo que podría **limitar su aplicabilidad** a otros contextos sanitarios y poblaciones con diferentes características genéticas, ambientales y socioeconómicas.  
- **Impacto del colapso de la Unión Soviética** (1991):  
  - Rusia atravesaba una **grave crisis económica y sanitaria** en los años 90.  
  - Escasez de insumos médicos, tecnología limitada y condiciones hospitalarias precarias podrían haber afectado la **calidad del tratamiento y la recopilación de datos**.  
  - Falta de acceso a medicamentos modernos y procedimientos avanzados podría haber influido en la **tasa de complicaciones** registradas.  

### 📉 **Valores faltantes en los datos (7.6%)**  
- La base de datos presenta un **7.6% de valores faltantes**, lo que puede afectar la **precisión y confiabilidad** del análisis.  
- Se requerirá aplicar **métodos de imputación o eliminación** para manejar estos datos de manera adecuada.  
- También se observa la ausencia de algunas columnas originales en el dataset, no solo por su numeración, sino también por los hallazgos. Por ejemplo, en los bloqueos falta el de **2º grado**. Asimismo, se identifican las siguientes ausencias en los grupos de variables:

````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, OrdinalEncoder, LabelEncoder
from sklearn.model_selection import StratifiedKFold, GridSearchCV, RandomizedSearchCV
from sklearn.metrics import f1_score, make_scorer, precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay, classification_report, accuracy_score
from sklearn.svm import SVC
from imblearn.over_sampling import SMOTE, ADASYN
from imblearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.decomposition import TruncatedSVD
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

df_prep = df.copy()

# En preparación de introducir los datos a nuestors modelos transformaremos nuestras variables a numeros de nuevo.

binaries_cat = [
    'has',
    'dm',
    'obsesidad',
    'antecedente_arritmia',
    'antecedente_bloqueo',
    'im_anterior',
    'im_lateral',
    'im_inferior',
    'im_posterior',
    'im_vd',
    'tratamiento_fibrinolitico'
]
categorics_features = [
    'sexo',
    #'muerte',
    #'complicaciones_mortales',
    'icc',
    'has',
    'dm',
    'obsesidad',
    'antecedente_arritmia',
    'antecedente_bloqueo',
    'im_anterior',
    'im_lateral',
    'im_inferior',
    'im_posterior',
    'im_vd',
    'tratamiento_fibrinolitico'
]
ordinal_features = [
    'cf_ultimo_ano',
    'cantidad_im_previos',
    'antecedente_angina',
    'enfermedad_coronaria_reciente',
    'tiempo_evolucion',
    'hallazgos_ecg_ingreso',
    'bloque_ecg_ingreso',
    'ritmo_ecg'
]

y_complicaciones_df = df_prep['complicaciones_mortales'].copy()
y_muerte_df = df_prep['muerte'].copy()
X_df= df_prep.drop(columns=['complicaciones_mortales', 'id', 'muerte']).copy()



preprocessorX = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), ['edad']),
        ('onehot', OneHotEncoder(handle_unknown='error'), categorics_features),
        ('ordinal', OrdinalEncoder(handle_unknown='error'), ordinal_features),
    ],
    remainder='drop'
)

labelEncoderMuerte = LabelEncoder()
labelEncoderComplicaciones = LabelEncoder()

X_svm = preprocessorX.fit_transform(X_df)

y_muerte_svm =  labelEncoderMuerte.fit_transform(y_muerte_df)

y_complicaciones_svm =  labelEncoderComplicaciones.fit_transform(y_complicaciones_df)

# Verificamos los NaN
print(f'Nan en X: {np.isnan(X_svm).any()}, y_muerte: {np.isnan(y_muerte_svm).any()} y_complicaciones: {np.isnan(y_complicaciones_svm).any()}')
print(f'Shape en X: {X_svm.shape}, y_muerte: {y_muerte_svm.shape} y_complicaciones: {y_complicaciones_svm.shape}')
````
